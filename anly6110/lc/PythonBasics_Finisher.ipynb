## Basic Data Types
**Python Basic Data Types**  
1. **Ints (Integers)** : Whole numbers, like `1`, `42`, `-10`.

## Boolean
**Boolean Values**
In Python, the two boolean values are:

- `True`

- `False`

These represent the truth values used in logical operations.
**Boolean Casting** You can convert other data types to boolean using the `bool()` function. The general rule is:
- **`bool(x)` returns `False`**  if `x` is considered "empty" or "falsey."

- **`bool(x)` returns `True`**  if `x` has a "truthy" value.

Here's how different values are cast to boolean:

1. **`bool(0)`**  → `False`
The number `0` is considered false.

2. **`bool(-1)`**  → `True`
Any non-zero number (positive or negative) is considered true.

3. **`bool([])`**  → `False`
An empty list `[]` is considered false.

4. **`bool({})`**  → `False`
An empty dictionary `{}` is also considered false.

5. **`bool(0.0)`**  → `False`
The float `0.0` is considered false.

6. **`bool(None)`**  → `False`
The special value `None` is always false.

### Example in Python:


```python
print(bool(0))       # False
print(bool(-1))      # True
print(bool([]))      # False
print(bool({}))      # False
print(bool(0.0))     # False
print(bool(None))    # False
```

### Summary of Truthy and Falsey Values:

- **Falsey values** : `0`, `0.0`, `None`, `[]`, `{}`, `""` (empty string), and `False`.

- **Truthy values** : Any non-zero numbers, non-empty collections (lists, tuples, sets, dictionaries), and `True`.

## String
**String Creation**  
1. **`str(123.3)`** : Converts a number to a string.
  - Example: `str(123.3)` → `'123.3'`

2. **Single (`'`) and Double (`"`) Quotes** : Both can be used to define strings.
  - Example: `""` or `''` are both valid empty strings.

3. **Concatenation** : Combining strings using `+`.
  - Example: `"123" + '20'` → `'12320'`
**Output Formatting with f-strings** :
1. **`f"{2.7:.3f}"`** : Formats a float to 3 decimal places.
  - Example: `f"{2.7:.3f}"` → `'2.700'`

2. **`f"{.27:.3%}"`** : Formats a number as a percentage with 3 decimal places.
  - Example: `f"{.27:.3%}"` → `'27.000%'`
**Slicing Strings** :
1. **`str[]`** : Access individual characters in a string.
  - Example: `str[1]` gives the second character (indexing starts at 0).

2. **`str[1:5]`** : Slices from index 1 to 4 (5 is exclusive).
  - Example: `"hello"[1:5]` → `'ello'`

3. **`str[2:]`** : Slices from index 2 to the end.
  - Example: `"hello"[2:]` → `'llo'`

4. **`str[:5]`** : Slices from the beginning to index 4.
  - Example: `"hello"[:5]` → `'hello'`

5. **`str[::2]`** : Slices every second character.
  - Example: `"hello"[::2]` → `'hlo'`
**String Length** :
- **`len(str)`** : Returns the length of the string.
  - Example: `len("hello")` → `5`
**String Formatting** :
- **`format()` method**  can also be used to format strings.
  - Example: `"{:.2f}".format(3.14159)` → `'3.14'`
**Multi-line Strings** :
- **Triple quotes**  are used to define multi-line strings.
  - Example: `'''long string here'''` or `"""multi-line string"""`
**Raw Strings** :
- **`r"\nhello\n"`** : Treats escape sequences (`\n`, `\t`, etc.) as literal characters.
  - Example: `r"\nhello\n"` → `'\nhello\n'`
**Common String Methods** :
1. **`strip()`** : Removes leading and trailing whitespace.
  - Example: `" hello ".strip()` → `'hello'`

2. **`upper()`** : Converts the string to uppercase.
  - Example: `"hello".upper()` → `'HELLO'`

3. **`lower()`** : Converts the string to lowercase.
  - Example: `"HELLO".lower()` → `'hello'`

4. **`startswith()`** : Checks if the string starts with a certain substring.
  - Example: `"hello".startswith("he")` → `True`

5. **`endswith()`** : Checks if the string ends with a certain substring.
  - Example: `"hello".endswith("lo")` → `True`

6. **`find()`** : Returns the index of the first occurrence of a substring, or `-1` if not found.
  - Example: `"hello".find("e")` → `1`, `"hello".find("x")` → `-1`

7. **`count()`** : Counts occurrences of a substring.
  - Example: `"hello".count("l")` → `2`

8. **`replace(old, new)`** : Replaces occurrences of a substring with another string.
  - Example: `"hello".replace("l", "z")` → `'hezzo'`

9. **`split(",")`** : Splits the string into a list based on a delimiter.
  - Example: `"a,b,c".split(",")` → `['a', 'b', 'c']`

10. **`split()`** : Splits the string by any whitespace.
  - Example: `"hello world".split()` → `['hello', 'world']`
**Joining Strings** :
- **`"--".join(list)`** : Joins a list of strings into a single string with a separator.
  - Example: `"--".join(['a', 'b', 'c'])` → `'a--b--c'`

### List
**Python List**
- Ordered, mutable, allows duplicate elements

- Allows different data types

- `list()` to create an empty list

- Example list: `[1, 2, "str", 23.4, []]`

- Access first element: `l[0]`

- `l.append(90)`: appends `90` to the end of the list

- `l.insert(1, "A")`: inserts `"A"` at index 1

- `l.remove(1)`: removes the first occurrence of `1`; raises an error if the element is not found

- `l.pop()`: removes and returns the last element

- `len(l)`: returns the length of the list

- `l.copy()`: returns a shallow copy of the list

- `l.sort(reverse=True)`: sorts the list in reverse order (modifies the original list)

- `sorted(l)`: returns a new sorted list (original list remains unchanged)

- `l.reverse()`: reverses the elements of the list (modifies the original list)

- Slicing: `l[1:5:2]` (start at index 1, go to index 5, step by 2)

- Operator: `in` (checks if an element is in the list)

- `for` loop to iterate over a list

- Repeating elements: `[1, 2] * 10` (repeats `[1, 2]` ten times)

**List Comprehensions** :
- `[item for item in l]`: creates a list from each item in `l`

- `[x * x for x in l]`: squares each item in `l`

- `[fun(x) for x in l]`: applies `fun(x)` to each item in `l`

- `[fun(x) for x in l if x > 3]`: applies `fun(x)` to each item in `l` that is greater than 3

- `[fun(x) for x in l for l in ll]`: applies `fun(x)` for each `x` in `l` and each `l` in `ll`
