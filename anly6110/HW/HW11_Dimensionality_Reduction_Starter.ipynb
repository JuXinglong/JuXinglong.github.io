The **wine dataset** contains data on 1,599 wine samples with the following 12 attributes:

1. **fixed acidity**: The acidity level, mean = 8.32, range = 4.6 to 15.9.

3. **citric acid**: Citric acid content, mean = 0.27, range = 0 to 1.0.

9. **pH**: Acidity or alkalinity, mean = 3.31, range = 2.74 to 4.01.
10. **sulphates**: Sulfate content, mean = 0.66, range = 0.33 to 2.0.
11. **alcohol**: Alcohol percentage, mean = 10.42%, range = 8.4 to 14.9.
12. **quality**: Wine quality rating (integer scale from 3 to 8), mean = 5.64.
