**Sales Data Analysis**

### Dataset Description

File: **SalesData.csv**

The dataset consists of **12,569** sales records, structured with **12 columns** containing details of customer purchases. Below is an overview of the dataset:

- **RowNO**: Unique identifier for each row.
- **SaleDate**: Date of the sale, formatted as MM/DD/YYYY.
- **CustomerID**: Unique ID assigned to each customer.
- **FirstName**: First name of the customer.
- **LastName**: Last name of the customer.
- **OrderID**: Unique ID for each order (multiple rows can share an OrderID if they are part of the same order).
- **ProductName**: Name of the product sold.
- **Category**: Category of the product (e.g., sneakers, boots, wedges).
- **Color**: Color of the product.
- **SalePrice**: Price of the product at the time of sale.
- **Quantity**: Number of units purchased in the transaction.
- **ManufacturerName**: Name of the manufacturer.
