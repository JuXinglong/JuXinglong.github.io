```python
# Q2
%%sql
select *
from sale
where FirstName='Lorene' and LastName='Charles'
order by SalePrice desc
;
```

```python
# Q3
# count() sum() avg() max() min()
%%sql
select count(*) as NRows,
       sum(SalePrice*Quantity) as TotalSales
from sale
;
```

```python
%%sql
select ManufacturerName,count(*) as NRows,
       sum(SalePrice*Quantity) as TotalSales
from sale
group by ManufacturerName
order by TotalSales desc
;
```

```python
# Q5
%%sql
select sum(Quantity) as TotalPairsSold,
       Category
from sale
group by Category
order by TotalPairsSold desc
;
```
