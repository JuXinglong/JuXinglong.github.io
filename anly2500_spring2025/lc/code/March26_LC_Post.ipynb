```python
gs = tips_df.groupby(by=["sex"]) #
```

```python
gs.groups.keys()
```

```python
i='Female'
s_df = gs.get_group(i)
#s_df # print(s_df)

print(f"Sex: {i}")
#
d=s_df['size'].sum()
print(f"Total number of customers served: {d}")

#
d=s_df['total_bill'].sum()
print(f"Sum of total bill: ${d:.3f}")

#
d = len(s_df)
print(f"Total number of orders: {d}")

ss_df=s_df.sort_values(by=["total_bill","tip"],ascending=[False,False])
print(ss_df)

ss_df.to_csv(f"{i}_tips.csv",index=False)
```

```python
gs = tips_df.groupby(by=['day'])
for i in gs.groups.keys():#["Male","Female"]:
  #i='Female'
  s_df = gs.get_group(i)
  #s_df # print(s_df)

  print(f"Sex: {i}")
  #
  d=s_df['size'].sum()
  print(f"Total number of customers served: {d}")

  #
  d=s_df['total_bill'].sum()
  print(f"Sum of total bill: ${d:.3f}")

  #
  d = len(s_df)
  print(f"Total number of orders: {d}")

  ss_df=s_df.sort_values(by=["total_bill","tip"],ascending=[False,False])
  print(ss_df)

  ss_df.to_csv(f"{i}_tips.csv",index=False)
```
