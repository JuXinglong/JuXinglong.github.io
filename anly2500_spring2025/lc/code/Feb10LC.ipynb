```python
# your real name and student id
import random # import module, the module name is randome
# a module is another file. This file name is called random
rn_ground_truth = random.randint(0,10)# . belong to, from will generate a random integer 0-10

print("Please take a guess (0-10): ")
userGuessStr=input()# string
userGuessInt=int(userGuessStr)# convert into integer

print(f"The groud truth is: {rn_ground_truth}")

if userGuessInt == rn_ground_truth:
  print("Correct guess!")
else:
  print("Wrong guess!")
```

```python
# module library (you are going to use codes
# in another file)
import random as r #

nCorrect=0
nWrong=0
nGiveup=0

totalPoints=0

nRounds=0

# 3: head
# 4: tail
# round 1
nRounds = nRounds+1 #
print(f"Round {nRounds}")
gtInt = r.randint(3,4) #
#print(gtInt)

print("Please take a guess (h,t,n):")
gStr=input()# string
if gStr=="h":
  gInt=3
elif gStr=="t":
  gInt=4
else:
  gInt=-1#

print(gtInt) # ground truth

if gInt==-1:
  print("Player gives up -1")
  nGiveup = nGiveup + 1
  totalPoints = totalPoints-1
elif gInt==gtInt:
  print("Correct guess +3")
  nCorrect = nCorrect + 1
  totalPoints=totalPoints+3
else:
  print("Wrong guess -5")
  nWrong =nWrong+1
  totalPoints=totalPoints-5

print(f"Correct: {nCorrect}")
print(f"Wrong: {nWrong}")
print(f"Give up: {nGiveup}")

print(f"Total points: {totalPoints}")
```
