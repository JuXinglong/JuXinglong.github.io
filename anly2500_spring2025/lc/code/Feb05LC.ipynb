```python
# list append() sort()
# your name and your student id

gList = [] # empty list. empty warehourse

print(gList)

# add new item into the list (warehouse)
gList.append(86.6) # . mean belongs to
print(gList)

gList.append(96.6)
gList.append(76.6)
print(gList)

print("Please input a new grade: ")
gStr = input() # string
gDbl = float(gStr) # convert to float (double)
gList.append(gDbl)

print(gList)
n = len(gList)
print("There are "+str(n)+" grades in this list")

# sort() ascending way
gList.sort()
print(gList)

gList.sort(reverse=True) # sort descending way
print(gList)
```
