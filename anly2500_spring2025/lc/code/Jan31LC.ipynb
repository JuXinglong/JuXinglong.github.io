```python
print("Please input your name: ")
nameStr = input() # string

print("Please input the age: ")
ageStr = input() # age string
ageInt = int(ageStr) # int() function call to conver the age string into a integer

print("Please input the mean amount: ")
amountStr = input() # string
amountDbl = float(amountStr)

b1 = ageInt<7 or ageInt>70

print(b1)

if b1 ==True:
  print("Congrats, "+nameStr+". You get the 50% discount.")
  print("The amount you need to pay is $"+str(amountDbl*0.5))
else:
  print("Sorry, "+nameStr+". You do not get the 50% discount.")
  print("You need to pay $"+ str(amountDbl))

```
