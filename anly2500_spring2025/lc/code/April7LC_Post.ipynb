```python
# Q7
# avg() max() sum() min() count()
%%sql
select ManufacturerName,
       sum(SalePrice*Quantity) as TotalRevenue,
       sum(Quantity) as PairsSold,
       avg(SalePrice) as AvgSalePrice
from sale
group by ManufacturerName
order by TotalRevenue desc
;
```

```python
# You need to do the lecture code prepare from April 9
#
# Q8
%%sql
select 	CustomerID,
        FirstName,
        LastName,
        sum(Quantity) as TotalPairs
from sale
group by CustomerID
order by TotalPairs desc
limit 1
;
```

```python
# Q9
%%sql
select OrderID,sum(Quantity) as TotalPairsSold
from sale
group by OrderID
having sum(Quantity)>9
order by TotalPairsSold desc
;
```
