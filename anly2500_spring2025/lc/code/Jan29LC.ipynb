```python
print("Please input your name: ")
nStr = input()

print("Are you a member of the store? (Yes/No)")
memStatusStr = input() # string

print("Please input the purchase amount: ")
amtStr = input() # string float

amtDbl = float(amtStr) # convert string to float

b = memStatusStr == "Yes" and amtDbl>50

print(b)

if b==True:
  print("Congrat,"+nStr+"! You get 10% discount")
  print("After discount, you need to pay $" + str(amtDbl*0.9))
else:
  print("Sorry, "+nStr+".You do not get 10% discount")
  print("You need to pay $" + str(amtDbl))
```
