```python
# if else (conditional) statement
# one if else, at most you will access to one code block
#
# if condition1:
#    code block 1
# else:
#   code blcok 2
g1 = 91
if g1>90:
   print("The grade is greater than 90")
   print("A")
else:
  print("Less than 90")
  print("Not A")
print("This is the next line if else")
```
