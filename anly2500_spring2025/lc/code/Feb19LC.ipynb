```python
# your name and your student id
# collect a few grades and then calculate the average

# counter/accumulator variables bank balance
sumGrade = 0 # the sum of the grades
nStu=0 # number of students or grades

print("Please input the number of students: ")
nStuStr = input() # string
nStu = int(nStuStr)

# for loop to collect the grades
for i in range(nStu):# range(nStu) range(3) 0, 1, 2
  print(f"Student {i+1} Grade:")
  gStr = input()
  gDbl = float(gStr)
  # deposit into the bank account
  sumGrade=sumGrade+gDbl

# calculate the average
# averageGrade = sumGrade/nStu
avgGrade = sumGrade/nStu
print(f"\n\nThe average grade is: {avgGrade:.2f}")
```
