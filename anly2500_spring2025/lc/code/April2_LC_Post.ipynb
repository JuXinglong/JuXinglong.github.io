```python
%%sql
select City as C,
       State as S,
       Population as P
from data
where State='Utah' or State='Arizona' or State='Texas'
order by Population desc
;
```

```python
%%sql
select City as C,
       State as S,
       Population as P,
       Population*1.1 as NewP
from data
where State in ('Utah','Texas','Arizona')
order by Population desc
limit 20
;
```

```python
# avg() sum() count() max() min()
%%sql
select State,
       sum(Population) as SUM_P,
       count(*) as NRows
from data
group by State
having count(*)>=20 and sum(Population)>10000000
order by NRows asc
;
```
