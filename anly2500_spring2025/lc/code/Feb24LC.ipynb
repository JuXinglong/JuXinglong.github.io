```python
# your name and your student id

# counter variables
nGuess=0 # number of guesses
totalP =0 # total points

# Generate a randome number
import random
gtNumber = random.randint(1,20) # ground truth
# print(gtNumber)
#
while True:# infinite loop
  #
  print("\n\nPlease take a guess (1-20):")
  userGuessStr = input()
  userGuessInt = int(userGuessStr)

  nGuess = nGuess+1

  if userGuessInt == gtNumber:
    # correct
    totalP=totalP+5
    print("Congrats! It is correct!")
    break# stop the loop, go to the summary
  else:
    # wrong
    totalP=totalP-1
    print("Sorry! It is wrong!")

# summary information
print(f"Total guesses: {nGuess}")
print(f"Total points: {totalP}")
```

```python
# cog unit price: 60
# gear: 200

# define counter/accumulator variables
# bank account
nOrder = 0
sumSub = 0 #sum of all orders' subtotal
nBelow=0 # below 1000, num of order whose subtotal <1000
nBetween=0 # 1000 - 2000
nAbove=0 # above 2000
nCogsSold=0
nGearsSold=0 # total gears sold


# collect the orders
while True:
  # collect one order
  nOrder = nOrder+1#

  print(f"Input N of Cogs in Order {nOrder}:")
  cogsStr=input()# string
  cogsInt=int(cogsStr)

  print(f"Input N of Gears in Order {nOrder}:")
  gearsStr=input()
  gearsInt=int(gearsStr)

  nCogsSold=nCogsSold+cogsInt
  nGearsSold=nGearsSold+gearsInt

  # subtotal in one order
  subtotal = 60*cogsInt+200*gearsInt

  #
  sumSub = sumSub+subtotal#

  # determine the subtotal level
  if subtotal<1000:
    print(f"Subtotal: ${subtotal:.2f}, BELOW 1000")
    nBelow=nBelow+1
  elif subtotal<=2000:
    print(f"Subtotal: ${subtotal:.2f}, BETWEEN")
    nBetween=nBetween+1
  else:
    print(f"Subtotal: ${subtotal:.2f}, ABOVE 2000")
    nAbove=nAbove+1



  print(f"\nAnother order? (yes/no)")
  userChoice=input()#string

  if userChoice=='no':
    break# stop the loop


# summary information
print(f"\n\nSummary information")
print(f"N orders: {nOrder}")
print(f"N cogs sold: {nCogsSold}")
print(f"N gears sold: {nGearsSold}")
print(f"Total sales: ${sumSub:.2f}")
print(f"Average subtotal: ${sumSub/nOrder:.2f}")

print(f"N Below: {nBelow}/{nOrder} {nBelow/nOrder:.2%}")
print(f"N Between:{nBetween}/{nOrder} {nBetween/nOrder:.2%}")
print(f"N Above: {nAbove}/{nOrder} {nAbove/nOrder:.2%}")
```
