```python
# your name and you student id (if you dont write your real name, -0.1 point off)
# while loop

# define counter variable
nStu = 0
tG = 0 # sum of the grades

print("Another grade? (y/n)")
userChoiceStr = input() # string

while userChoiceStr=="y":
  # collect one student's grade
  nStu=nStu +1
  print(f"Input Student {nStu} grade: ")
  gStr = input()
  gDbl = float(gStr)

  tG = tG +gDbl

  # ask the question, another student's grade
  print("\nAnother grade? (y/n)")
  userChoiceStr = input()

print("\n\nSummary Information")
print(f"Total number ofs students: {nStu}")
print(f"Average grade: {tG/nStu:.2f}")
```

```python
# student grades

# counter variables
nS = 0
tG = 0 #
nA = 0
nB = 0
nC =0
nD =0
nF =0

# collect the grades
while True: # infinite loop
  # collect one student grade
  nS=nS+1 #
  print(f"Input student {nS} grade:")
  gStr=input()# grade
  g=float(gStr)

  tG = tG+g #

  # determine letter grade
  if g>=90:
    lg="A"
    nA=nA+1
  elif g>=80:
    lg='B'
    nB=nB+1
  elif g>=70:
    lg='C'
    nC=nC+1
  elif g>=60:
    lg="D"
    nD=nD+1
  else: # <60
    lg='F'
    nF=nF+1

  print(f"Your letter grade is {lg}.")





  print("\nAnother student's grade: (yes/no)")
  userChoiceStr=input() #string
  if userChoiceStr=="no":
    break

# summary information
print("\n\nsummary information")

print(f"N Students: {nS}")
print(f"Average grade: {tG/nS:.2f}")
print(f"N A: {nA}/{nS} ({nA/nS:.2%})")
print(f"N B: {nA}/{nS} ({nA/nS:.2%})")
```
