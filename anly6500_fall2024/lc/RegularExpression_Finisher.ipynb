## String

In Python, there are multiple ways to define string literals, and the choice often depends on what you need to achieve. Let's break down each of the types of string definitions you mentioned:
1. **Basic String** A basic string can be created using either single (`'...'`) or double quotes (`"..."`). Both are equivalent and used to enclose short strings or when one form of quote allows you to avoid escaping the other.
**Example:**


```python
s1 = 'Hello, World!'
s2 = "Hello, World!"
```

You can use them interchangeably, but they can be useful when you need to include quotation marks inside the string without using escape characters:


```python
s1 = 'She said "hello"'
s2 = "It's a sunny day"
```
2. **Raw String (`r"..."` or `r'...'`)** A raw string is prefixed with `r` or `R` and treats backslashes (`\`) as literal characters, meaning they are not interpreted as escape characters. This is especially useful in regex patterns, file paths, or Windows paths where backslashes are common.

**Example:**


```python
raw_string = r"C:\Users\Name\Documents"
print(raw_string)  # Output: C:\Users\Name\Documents
```
Without the `r` prefix, you'd need to double each backslash:

```python
regular_string = "C:\\Users\\Name\\Documents"
print(regular_string)  # Output: C:\Users\Name\Documents
```

Similarly, in regex:


```python
pattern = r"\d+"  # Raw string, \d is treated as a literal part of the regex
```

3. **Triple-quoted String (`'''...'''` or `"""..."""`)** Triple quotes allow you to write multi-line strings or strings that include both single and double quotes without escaping them. Triple quotes can either use single (`'''...'''`) or double quotes (`"""..."""`), and both are functionally the same.
**Example:**


```python
multi_line_string = '''This is
a multi-line
string.'''
print(multi_line_string)

# Output:
# This is
# a multi-line
# string.
```

Embedding quotes without escaping:


```python
complex_string = """This string has 'single quotes' and "double quotes" without escaping."""
```

Triple quotes are often used in docstrings (comments that serve as documentation for functions, classes, or modules) because they allow for clean multi-line comments.
4. **Differences and Use Cases**  
- **`r"..."`**  or ****`r"..."`**  or `r'...'`** : Use when you want to treat backslashes literally, without escaping, typically for regex patterns or file paths.

- **`'...'`**  or ****`'...'`**  or `"..."`** : Use for simple single-line strings or when you need to include one type of quote within the other without escaping.

- **`'''...'''`**  or ****`'''...'''`**  or `"""..."""`** : Use for multi-line strings or complex strings that might include both single and double quotes without needing to escape them. Also commonly used for docstrings.

5. **Practical Examples:**

**Regular String**


```python

```

**Raw String (important for regex or file paths)**


```python
file_path = r"C:\new_folder\test.txt"

```

**Triple-quoted String (for multiline text)**


```python

```


**Summary:**

- **`r"..."`**  (raw strings) are useful when dealing with backslashes (e.g., file paths, regex).

- **Single/Double quoted strings**  are best for simpler cases.

- **Triple quotes**  are great for multi-line strings or when you need to include both single and double quotes inside the string without escaping.

## Regular Expression

<a href="https://docs.python.org/3/library/re.html">**Regular Expression Tutorial**</a>

Regular expressions (regex) in Python. Regex is a powerful tool for searching, matching, and manipulating strings based on patterns.

1. **To use regex in Python, you need to import the `re` module.**

```python
import re
```

2. **Basic Patterns**  
- **`.`**  - Matches any single character except newline.

- **`^`**  - Anchors the pattern at the start of the string.

- **`$`**  - Anchors the pattern at the end of the string.

- **`[]`**  - Matches any character inside the square brackets.

- **`|`**  - Logical OR between patterns.

3. **Special Sequences**  
- **`\d`**  - Matches any digit (equivalent to [0-9]).

- **`\D`**  - Matches any non-digit character.

- **`\w`**  - Matches any alphanumeric character (equivalent to [a-zA-Z0-9_]).

- **`\W`**  - Matches any non-alphanumeric character.

- **`\s`**  - Matches any whitespace character.

- **`\S`**  - Matches any non-whitespace character.

4. **Quantifiers**  
- **`*`**  - Matches 0 or more repetitions of the preceding character.

- **`+`**  - Matches 1 or more repetitions.

- **`?`**  - Matches 0 or 1 repetition.

- **`{n}`**  - Matches exactly `n` repetitions.

- **`{n,m}`**  - Matches between `n` and `m` repetitions.

5. **Functions in the `re` Module** `re.match()`
This function checks for a match only at the beginning of the string.


```python
pattern = r'^Hello'
string = 'Hello World'
match = re.match(pattern, string)
print(match.group())  # Output: Hello
```
`re.search()`
Searches the string for a match, and returns the first occurrence.


```python
pattern = r'World'
string = 'Hello World'
search = re.search(pattern, string)
print(search.group())  # Output: World
```
`re.findall()`
Returns a list of all matches found in the string.


```python
pattern = r'\d+'
string = 'There are 12 apples and 34 oranges'
all_matches = re.findall(pattern, string)
print(all_matches)  # Output: ['12', '34']
```
`re.sub()`
Replaces occurrences of the pattern in the string with another string.


```python
pattern = r'apples'
replacement = 'bananas'
string = 'I love apples'
result = re.sub(pattern, replacement, string)
print(result)  # Output: I love bananas
```
`re.split()`
Splits the string by occurrences of the pattern.


```python
pattern = r'\s+'
string = 'Split this sentence into words'
split_result = re.split(pattern, string)
print(split_result)  # Output: ['Split', 'this', 'sentence', 'into', 'words']
```

6. **Example Use Cases**

**Example 1: Validate an Email**


```python
email = 'test@example.com'
pattern = r'^\w+@\w+\.\w+$'
if re.match(pattern, email):
    print('Valid email')
else:
    print('Invalid email')
```

**Example 2: Extract Dates from Text **


```python
text = 'The event is on 2024-10-15 and 2024-11-01.'
pattern = r'\d{4}-\d{2}-\d{2}'
dates = re.findall(pattern, text)
print(dates)  # Output: ['2024-10-15', '2024-11-01']
```


7. **Compiling Regular Expressions**
For performance improvement when using the same regex multiple times, you can compile the regex:


```python
pattern = re.compile(r'\d+')
result = pattern.findall('There are 12 apples and 34 oranges')
print(result)  # Output: ['12', '34']
```
8. **Flags**
You can modify regex behavior using flags:

- **`re.IGNORECASE` (or `re.I`)**  - Makes the pattern case-insensitive.

- **`re.MULTILINE` (or `re.M`)**  - Allows `^` and `$` to match the start and end of each line in the string.

Example with flag:


```python
pattern = r'^hello'
string = 'Hello world'
match = re.search(pattern, string, re.IGNORECASE)
print(match.group())  # Output: Hello
```
9. **Escaping Special Characters** If you want to match a character that is normally a special character in regex (like `.` or `*`), use a backslash `\` to escape it.Example: Matching a literal dot (`.`):

```python
pattern = r'\.'
string = 'Find the dot in this sentence.'
match = re.search(pattern, string)
print(match.group())  # Output: .
```

**Summary**

- Use `re.match()` to check from the beginning of the string.

- Use `re.search()` to find the first match.

- Use `re.findall()` to find all matches.

- Use `re.sub()` to replace patterns.

- Use `re.split()` to split based on a pattern.


1. Finding All Matches
If you want to find **all occurrences**  of a pattern, you can use `re.finditer()` instead of `re.search()`. This method returns an iterator that yields match objects for all matches in the string.
Here's how you can use it:


```python
import re

# Example string with multiple occurrences
text = "This is a sample text. Another sample is here."

# Regular expression pattern to search for the word 'sample'
pattern = r'sample'

# Use finditer() to find all matches
matches = re.finditer(pattern, text)

# Iterate through all matches
for match in matches:
    start, end = match.span()
    print(f"Match found from index {start} to {end}")
```

This will output:


```css
Match found from index 10 to 16
Match found from index 32 to 38
```

2. Extracting Start, End, and the Matched Text

You can also extract the start and end positions along with the matched text:


```python
import re

text = "The quick brown fox jumps over the lazy fox."

# Pattern to match the word 'fox'
pattern = r'fox'

matches = re.finditer(pattern, text)

# Iterate over matches and print match, start, and end positions
for match in matches:
    start, end = match.span()
    matched_text = match.group()
    print(f"Matched: '{matched_text}' from index {start} to {end}")
```

Output:


```csharp
Matched: 'fox' from index 16 to 19
Matched: 'fox' from index 40 to 43
```

3. Matching with Capture Groups
If you have patterns with **capture groups** , you can still get the start and end positions of both the overall match and individual groups.
Example with capture groups:


```python
import re

# Example string
text = "John's phone number is 123-456-7890."

# Pattern with capture groups (area code, first 3 digits, last 4 digits)
pattern = r'(\d{3})-(\d{3})-(\d{4})'

match = re.search(pattern, text)

if match:
    # Full match span
    print(f"Full match from index {match.start()} to {match.end()}: {match.group()}")
    
    # Individual group spans
    print(f"Area code from index {match.start(1)} to {match.end(1)}: {match.group(1)}")
    print(f"First 3 digits from index {match.start(2)} to {match.end(2)}: {match.group(2)}")
    print(f"Last 4 digits from index {match.start(3)} to {match.end(3)}: {match.group(3)}")
```

Output:


```sql
Full match from index 21 to 33: 123-456-7890
Area code from index 21 to 24: 123
First 3 digits from index 25 to 28: 456
Last 4 digits from index 29 to 33: 7890
```

4. Handling Non-Matches
If the pattern does not match any part of the string, the methods return `None`, so it’s important to check for this before accessing match details:

```python
import re

# String with no matching pattern
text = "No phone number here."

pattern = r'\d{3}-\d{3}-\d{4}'

match = re.search(pattern, text)

if match:
    print(f"Match found from index {match.start()} to {match.end()}: {match.group()}")
else:
    print("No match found.")
```

5. Summary of Useful Methods

- `match.start()`: Returns the start position of the match.

- `match.end()`: Returns the end position (exclusive).

- `match.span()`: Returns a tuple (start, end) with the start and end positions.

- `match.group()`: Returns the entire matched string.

- `match.group(n)`: Returns the `n`th capture group.
