## Introduction to Databases, Primary Keys, and Foreign Keys

#### What is a Database?
A **database**  is an organized collection of data that is stored electronically and can be easily accessed, managed, and updated. In a database, data is typically organized into **tables** , where each table represents a specific entity, like "Customers" or "Products." Each table is made up of **columns**  (attributes) and **rows**  (records).For example, consider a table called **Customers** :

| CustomerID | Name | Email |
| --- | --- | --- |
| 1 | Alice | alice@email.com |
| 2 | Bob | bob@email.com |

Each row represents a record, and each column represents a piece of information about that record.

#### What is a Primary Key?
A **primary key**  is a column or a set of columns in a database table that uniquely identifies each record in that table. It ensures that no two records in the table have the same primary key value, which guarantees uniqueness.In the **Customers**  table above, the **CustomerID**  field can be a primary key because each value in the column is unique for each customer. This allows the database to quickly identify, retrieve, and organize information for each customer.
Characteristics of a Primary Key:

1. **Unique** : No two records can have the same primary key value.

2. **Non-null** : The primary key field cannot be left empty.

3. **Unchanging** : The value of the primary key should remain constant.

#### What is a Foreign Key?
A **foreign key**  is a column or set of columns in one table that refers to the **primary key**  of another table. The foreign key creates a relationship between two tables, enabling cross-referencing of data. It ensures referential integrity, meaning that the data is consistent across related tables.For example, consider a second table called **Orders** :

| OrderID | OrderDate | CustomerID |
| --- | --- | --- |
| 101 | 2023-09-21 | 1 |
| 102 | 2023-09-22 | 2 |

In this case, the **CustomerID**  column in the **Orders**  table is a **foreign key**  that refers to the **CustomerID**  in the **Customers**  table. This creates a relationship between customers and their orders, where each order can be linked to a specific customer.
Characteristics of a Foreign Key:

1. **Links Tables** : It connects rows in one table to rows in another.

2. **Referential Integrity** : It ensures that the value in the foreign key column exists in the primary key column of the related table.

#### How Do Primary and Foreign Keys Work Together?
Primary and foreign keys work together to define relationships between tables in a relational database. By connecting tables, they allow complex queries that span multiple tables. For instance, you can use the **CustomerID**  foreign key in the **Orders**  table to retrieve all orders placed by a specific customer.
This relationship helps to keep the database organized, prevent duplication of data, and maintain data consistency.

### Summary

- A **database**  organizes data into tables, each with rows and columns.

- A **primary key**  uniquely identifies records in a table.

- A **foreign key**  links two tables together by referring to the primary key of another table, establishing a relationship.

- Together, primary and foreign keys ensure data integrity and enable powerful data queries across multiple tables.

This forms the foundation of how relational databases function, enabling efficient data management, retrieval, and relationship building.

## Syntax of Select

1. **Order of Clauses Is Set:**  
  - The order of SQL clauses is crucial. Changing the order will result in an error. SQL clauses must follow this strict order:
    1. `SELECT`

    2. `FROM`

    3. `WHERE`

    4. `GROUP BY`

    5. `HAVING`

    6. `ORDER BY`
If you mix the order (e.g., placing `ORDER BY` before `GROUP BY`), the query will fail to execute.

2. **Capitalize Commands for Readability:**  
  - SQL commands are case-insensitive, meaning it doesn't matter if they are written in uppercase or lowercase. However, it is a good practice to **capitalize SQL commands**  (`SELECT`, `FROM`, `WHERE`, etc.) to improve readability and make the code more standardized.

Example of readable SQL:


```sql
SELECT name, age
FROM students
WHERE age > 18
ORDER BY age DESC;
```
Additionally, **indentation**  and writing each clause on a new line makes the query much more readable and easier to debug.

`WHERE` Clause in SQLThe `WHERE` clause is used to filter records in a SQL query based on certain conditions. It is essential in retrieving only the rows that meet specific criteria.

**Basic Syntax of Select**


```sql
SELECT column1, column2, ...
FROM table_name
WHERE condition;
```
Common Operators for the `WHERE` Clause

1. **Comparison Operators:**  
- `=`: Equal to

- `<>`: Not equal to (or `!=` in some databases)

- `>`: Greater than

- `<`: Less than

- `>=`: Greater than or equal to

- `<=`: Less than or equal to

**Example:**


```sql
SELECT *
FROM employees
WHERE salary > 50000;
```

This query selects employees with a salary greater than 50,000.
2. **Logical Operators:**  
- `AND`: Combines multiple conditions; both must be true.

- `OR`: Combines multiple conditions; either one must be true.

- `NOT`: Negates a condition.
Example with `AND` and `OR`:

```sql
SELECT *
FROM employees
WHERE salary > 50000 AND department = 'IT';
```

This retrieves IT department employees with a salary greater than 50,000.


```sql
SELECT *
FROM employees
WHERE department = 'IT' OR department = 'HR';
```

This selects employees who belong to either the IT or HR departments.
Example with `NOT`:

```sql
SELECT *
FROM employees
WHERE NOT department = 'HR';
```

This selects all employees except those in the HR department.


3. **`BETWEEN`:**
This operator is used to filter values within a specific range.

**Example:**


```sql
SELECT *
FROM employees
WHERE salary BETWEEN 30000 AND 60000;
```

This query retrieves employees with salaries between 30,000 and 60,000.

4. **`IN`:** The `IN` operator allows filtering based on multiple values.

**Example:**


```sql
SELECT *
FROM employees
WHERE department IN ('IT', 'HR', 'Sales');
```

This query retrieves employees from the IT, HR, or Sales departments.

5. **`LIKE`:** The `LIKE` operator is used to search for patterns in text columns, often with wildcards:
- `%` matches any sequence of characters.

- `_` matches a single character.

**Example:**


```sql
SELECT *
FROM employees
WHERE name LIKE 'J%';
```

This retrieves employees whose name starts with "J."

6. **`NULL`:** The `IS NULL` and `IS NOT NULL` operators check for `NULL` values.

**Example:**


```sql
SELECT *
FROM employees
WHERE manager_id IS NULL;
```

This query retrieves employees who do not have a manager.


---


**Combining Multiple Conditions**

You can combine different operators and conditions in a `WHERE` clause using logical operators (`AND`, `OR`, `NOT`).

**Example:**


```sql
SELECT *
FROM employees
WHERE department = 'Sales'
  AND salary > 40000
  AND hire_date BETWEEN '2020-01-01' AND '2022-12-31';
```

This query retrieves Sales department employees hired between 2020 and 2022 with a salary greater than 40,000.

### Example Query Combining Functions:

Let's combine these functions in a query:


```sql

       SUM(salary) AS total_salary,
       AVG(salary) AS average_salary,
```

- Highest and lowest salaries

- The number of employees in each department



### SQL Query Structure Tutorial

SQL (Structured Query Language) is used to manage and manipulate relational databases. A typical SQL query follows a specific structure with different clauses that serve distinct purposes. Here's a breakdown of each part of a query, which aligns with the image structure:
1. `SELECT`The `SELECT` statement is used to specify the columns or expressions that you want to retrieve from a database table.

Syntax:


```sql
SELECT column1, column2, ...
FROM table_name;
```

Example:


```sql
SELECT name, age FROM students;
```
This query retrieves the `name` and `age` columns from the `students` table.

2. `FROM`The `FROM` clause specifies the table from which the data is retrieved.

Syntax:


```sql
FROM table_name;
```

Example:


```sql
SELECT name, age
FROM students;
```
Here, data is selected from the `students` table.

3. `WHERE`The `WHERE` clause filters the records based on specific conditions. Only the rows that meet the conditions will be included in the result set.

Syntax:


```sql
WHERE condition;
```

Example:


```sql
SELECT name, age
FROM students
WHERE age > 18;
```
This query selects the `name` and `age` of students who are older than 18.

4. `GROUP BY`The `GROUP BY` clause is used to group rows that have the same values in specified columns. Often, this is used with aggregate functions (`COUNT()`, `SUM()`, `AVG()`, etc.).

#### Syntax:


```sql
GROUP BY column1, column2, ...;
```

Example:


```sql
SELECT department, COUNT(*)
FROM employees
GROUP BY department;
```
This groups the employees by their `department` and counts the number of employees in each department.

5. `HAVING`The `HAVING` clause is used to filter records after the `GROUP BY` operation. It is similar to `WHERE`, but `WHERE` cannot be used with aggregate functions.

Syntax:


```sql
HAVING condition;
```

Example:


```sql
SELECT department, COUNT(*)
FROM employees
GROUP BY department
HAVING COUNT(*) > 10;
```

This query selects departments that have more than 10 employees.

6. `ORDER BY`The `ORDER BY` clause is used to sort the result set in either ascending (`ASC`) or descending (`DESC`) order based on one or more columns.

Syntax:


```sql
ORDER BY column1 [ASC|DESC], column2 [ASC|DESC], ...;
```

Example:


```sql
SELECT name, age
FROM students
WHERE age > 18
ORDER BY age DESC;
```

This sorts the students who are older than 18 by age in descending order.


---


Complete Example:


```sql
SELECT department, COUNT(*) AS total_employees
FROM employees
WHERE salary > 50000
GROUP BY department
HAVING COUNT(*) > 10
ORDER BY total_employees DESC;
```

In this example:

- We are selecting departments and counting the number of employees.

- The `WHERE` clause filters employees with a salary greater than 50,000.

- The `GROUP BY` groups the results by department.

- The `HAVING` clause filters the grouped results to show only departments with more than 10 employees.

- The `ORDER BY` sorts the final results in descending order based on the number of employees.


---


Practice

Try to build your own SQL queries using this structure. Start with a basic `SELECT` and gradually add the `WHERE`, `GROUP BY`, `HAVING`, and `ORDER BY` clauses based on your data and analysis requirements.

## CREATE TABLE

1. **Basic `CREATE TABLE` Syntax** To create a table, you use the `CREATE TABLE` statement. You must define the table name and the columns along with their data types.

```sql
CREATE TABLE table_name (
    column1 datatype constraint,
    column2 datatype constraint,
    column3 datatype constraint,
    ...
);
```

- `table_name`: The name of the table.

- `column1, column2, column3`: The names of the columns in the table.

- `datatype`: Specifies the data type of the column (e.g., `INT`, `VARCHAR`, `DATE`, etc.).

- `constraint`: Optional. You can add constraints like `NOT NULL`, `UNIQUE`, `PRIMARY KEY`, etc.

2. **Example: Creating a Table**
Let's create a table to store information about employees:


```sql
CREATE TABLE Employees (
    EmployeeID INT PRIMARY KEY,
    FirstName VARCHAR(50) NOT NULL,
    LastName VARCHAR(50) NOT NULL,
    BirthDate DATE,
    Email VARCHAR(100) UNIQUE,
    HireDate DATE,
    Salary DECIMAL(10, 2)
);
```

- `EmployeeID`: Primary key of the table (each employee will have a unique ID).

- `FirstName` and `LastName`: Both are `VARCHAR(50)` and cannot be null.

- `BirthDate`: A `DATE` column.

- `Email`: A `VARCHAR(100)` with a `UNIQUE` constraint, meaning no two employees can have the same email address.

- `HireDate`: Another `DATE` column.

- `Salary`: A `DECIMAL(10, 2)` type, which can store up to 10 digits with 2 decimal places.

3. **Adding Constraints**
You can add constraints to ensure data integrity. Some common constraints include:

- **`PRIMARY KEY`** : Uniquely identifies each record.

- **`NOT NULL`** : Ensures that a column cannot have `NULL` values.

- **`UNIQUE`** : Ensures all values in a column are unique.

- **`DEFAULT`** : Specifies a default value if none is provided.

- **`CHECK`** : Ensures all values in a column meet a specific condition.

Example:


```sql
CREATE TABLE Products (
    ProductID INT PRIMARY KEY,
    ProductName VARCHAR(100) NOT NULL,
    Price DECIMAL(10, 2) CHECK (Price > 0),
    Stock INT DEFAULT 0
);
```

In this table:

- `Price` must be greater than zero (`CHECK` constraint).

- If `Stock` is not provided, it defaults to 0 (`DEFAULT` constraint).

4. **`AUTO_INCREMENT`/`SERIAL`** For tables that require an auto-incrementing column (like `EmployeeID`), you can use `AUTO_INCREMENT` in MySQL or `SERIAL` in PostgreSQL.
Example for MySQL:


```sql
CREATE TABLE Employees (
    EmployeeID INT AUTO_INCREMENT PRIMARY KEY,
    FirstName VARCHAR(50) NOT NULL,
    LastName VARCHAR(50) NOT NULL
);
```

In PostgreSQL, you can use:


```sql
CREATE TABLE Employees (
    EmployeeID SERIAL PRIMARY KEY,
    FirstName VARCHAR(50) NOT NULL,
    LastName VARCHAR(50) NOT NULL
);
```
5. **Foreign Keys**
You can use foreign keys to create relationships between tables.

Example:


```sql
CREATE TABLE Departments (
    DepartmentID INT PRIMARY KEY,
    DepartmentName VARCHAR(100) NOT NULL
);

CREATE TABLE Employees (
    EmployeeID INT PRIMARY KEY,
    FirstName VARCHAR(50) NOT NULL,
    LastName VARCHAR(50) NOT NULL,
    DepartmentID INT,
    FOREIGN KEY (DepartmentID) REFERENCES Departments(DepartmentID)
);
```
In this example, `DepartmentID` in the `Employees` table is a foreign key that references the `DepartmentID` in the `Departments` table.

6. **Altering a Table** If you need to modify an existing table (add, modify, or drop columns), you can use the `ALTER TABLE` statement.
Example to add a new column:


```sql
ALTER TABLE Employees ADD PhoneNumber VARCHAR(15);
```

7. **Dropping a Table**
To delete an existing table, you can use:


```sql
DROP TABLE table_name;
```

Example:


```sql
DROP TABLE Employees;
```

## INSERT and DELETE

1. **SQL to Add a Row (INSERT)** To add a row into a table, you use the `INSERT INTO` statement.

```sql
INSERT INTO table_name (column1, column2, column3, ...)
VALUES (value1, value2, value3, ...);
```

Example:
If you have a table called `employees` with columns `employee_id`, `first_name`, `last_name`, and `email`, and you want to add a new employee:

```sql
INSERT INTO employees (employee_id, first_name, last_name, email)
VALUES (101, 'John', 'Doe', 'john.doe@example.com');
```
2. **SQL to Delete a Row (DELETE)** To delete a row, you use the `DELETE FROM` statement along with a condition (`WHERE` clause) to specify which row(s) to delete. Be cautious when using the `DELETE` statement without a `WHERE` clause, as it will delete all rows from the table.

```sql
DELETE FROM table_name
WHERE condition;
```

Example:
If you want to delete the row where `employee_id` is `101` from the `employees` table:

```sql
DELETE FROM employees
WHERE employee_id = 101;
```

## SQL datatypes

SQL datatypes define the kind of data that can be stored in a database column. Here is a summary of common SQL datatypes:

1. **Numeric Datatypes**  
- **INT** : Integer value, used for whole numbers (e.g., 1, 100, -50).

- **FLOAT** : Floating point number with approximate precision.

- **DECIMAL (M, D) / NUMERIC (M, D)** : Fixed-point numbers with exact precision, where `M` is the total number of digits and `D` is the number of digits after the decimal point.

- **BIGINT** : Larger range of integer values.

- **SMALLINT** : Smaller range of integer values.

- **TINYINT** : Very small range of integer values.

2. **String (Character) Datatypes**  
- **CHAR(n)** : Fixed-length character string. `n` defines the length of the string.

- **VARCHAR(n)** : Variable-length string. `n` defines the maximum length of the string.

- **TEXT** : Variable-length text, often used for large blocks of text.

- **BLOB** : Binary large object, used for storing binary data like images or files.

3. **Date and Time Datatypes**  
- **DATE** : Date in the format `YYYY-MM-DD`.

- **TIME** : Time in the format `HH:MM:SS`.

- **DATETIME** : Combination of date and time in the format `YYYY-MM-DD HH:MM:SS`.

- **TIMESTAMP** : Stores date and time but is typically used for tracking record changes.

- **YEAR** : Stores a year in either two digits or four digits format.

4. **Boolean Datatype**  

- **BOOLEAN** : Stores TRUE or FALSE values. Often represented internally as `TINYINT` with `0` for FALSE and `1` for TRUE.

5. **Binary Datatypes**  

- **BINARY(n)** : Fixed-length binary data.

- **VARBINARY(n)** : Variable-length binary data.

6. **Miscellaneous Datatypes**  

- **ENUM** : A string object with a predefined set of values (e.g., `ENUM('small', 'medium', 'large')`).

- **SET** : Similar to ENUM, but allows multiple values from the predefined set to be stored.

- **JSON** : Stores JSON formatted data (available in some database systems like MySQL).

- **UUID** : Universally unique identifier, used for unique key generation.

**Notes:**

- The choice of datatype affects the performance and storage requirements of the database, so it's important to select the most appropriate type based on the data being stored.
