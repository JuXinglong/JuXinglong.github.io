```python
fig = plt.figure(figsize=(15,4)) # inch

# 1 row 3 cols
ax1 = plt.subplot(1,3,1)

x=df1["total_bill"]
y=df1["tip"]

sc1 = plt.scatter(x,y,c=y,cmap=plt.cm.jet,label="Original data",marker="P",s=x)

cb1 = plt.colorbar(sc1)
cb1.set_label("Tip")

plt.xlabel("Total bill",fontsize=8)
plt.ylabel("Tip")
plt.title("Tip vs total bill")

x=[0,50]
y=[0,8]
plt.plot(x,y,label="Trending line",c="#289911",lw=5,ls="dashed")

plt.legend()

#
ax2 = plt.subplot(1,3,2)

plt.bar(vc.index,vc.values,color="green",width=0.6) # 0-1

# ax3
ax3 = plt.subplot(1,3,3)
# histogram
plt.hist(df1["tip"],bins=10)

# png jpg svg pdf
plt.savefig("fig1.png",dpi=600)

plt.show()
```

```python
df2 = pd.read_csv("CA_Housing_Data.csv")
df2
```

```python
df2.columns
```

```python
fig = plt.figure(figsize=(10,4))

# 1 row 2 columns
ax1 = plt.subplot(1,2,2)

x=df2["longitude"]
y=df2["latitude"]

pop = df2["population"]
value = df2["median_house_value"]

sc1 = plt.scatter(x,y,s=pop/50,c=value,cmap=plt.cm.jet)
cb1 = plt.colorbar(sc1)
cb1.set_label("Median house value")

ax2 = plt.subplot(1,2,1)
plt.hist(value)

plt.savefig("fig2.jpg",dpi=600)

plt.show()
```
