```python
fig = plt.figure(figsize=(15,4))

# 1 row 3 columns
ax1 =plt.subplot(1,3,1)
x=df1["total_bill"]
y=df1["tip"]

sc1 = plt.scatter(x,y,marker="P",c=y,cmap=plt.cm.jet,s=x)
cb1 = plt.colorbar(sc1)
cb1.set_label("Tip")

plt.xlabel("Total bill")
plt.ylabel("Tip")
plt.title("Tip vs total bill",fontsize=8)

# ax2 distribution histogram
ax2 = plt.subplot(1,3,2)
plt.hist(df1["tip"],bins=10)

#ax3 bar
ax3 = plt.subplot(1,3,3)
plt.bar(vc.index,vc.values,width=0.7) # 0-1

plt.savefig("fig1.png",dpi=600)

plt.show()
```
